```python
import re
```

## 1. Square brackets `[ ]`

### `[mM]` : one of the characters inside
Matches *Mary* or *mary*.

```python
text = "Mary Ann stopped by mary's house"
print(re.findall(r"[mM]ary", text))
```

### `[abc]` : a, b or c

```python
text = "In uomini, in soldati"
print(re.findall(r"[abc]", text))
```

### `[A-Z]` : one uppercase letter (a range)

```python
text = "we should call it Drenched Blossoms"
print(re.findall(r"[A-Z]", text))
```

### `[a-z]` : one lowercase letter

```python
text = "Hello World 10 $ 3.2"
print(re.findall(r"[a-z]", text))
```

### `[0-9]` : one digit

```python
text = "Chapter 1: Down the Rabbit Hole, page 25"
print(re.findall(r"[0-9]", text))
```

## 2. Negation `[^ ]`

### `[^A-Z]` : not an uppercase letter
The caret `^` is the first symbol after `[`.

```python
text = "Oyfn PRI"
print(re.findall(r"[^A-Z]", text))
```

### `[^Ss]` : neither S nor s

```python
text = "Sara sees"
print(re.findall(r"[^Ss]", text))
```

### `[e^]` : e or ^
The caret is not first, so it is just a caret.

```python
text = "look up ^ here"
print(re.findall(r"[e^]", text))
```

## 3. Counting, optionality and wildcards

### `?` : zero or one of the previous character

```python
text = "color colour"
print(re.findall(r"colou?r", text))
```

### `?` : an optional s
Note: the book writes `koala?`, but the `?` must come after the **s**.

```python
text = "one koala, two koalas"
print(re.findall(r"koalas?", text))
```

### `*` : zero or more of the previous character
`baaa*!` = b, then aa, then zero or more a's, then !

```python
text = "b! ba! baa! baaa! baaaaa!"
print(re.findall(r"baaa*!", text))
```

### `+` : one or more of the previous character

```python
text = "b! ba! baa! baaa! baaaaa!"
print(re.findall(r"baa+!", text))
```

### `+` : a number (one or more digits)

```python
text = "Room 7 and room 2048"
print(re.findall(r"[0-9]+", text))
print("\n")
print(re.findall(r"[0-9]*", text))

# what the difference between them
```

### `{n}` : exactly n times
`x{3}` = exactly three x's.

```python
text = "axxxz axxz axxxxz"
print(re.findall(r"ax{3}z", text))
print("\n")

text = "07xxx 07xxx-xxx-xxx xx xxxxz"
print(re.findall(r"07x{3}", text))
```

### `.` : any single character

```python
text = "big bag bug b g"
print(re.findall(r"b.g", text))
```

### `.*` : any string
A line with *rose* twice.

```python
text = "a rose is a rose"
print(re.findall(r"rose.*rose", text))
```

## 4. Anchors and boundaries

### `^` : start of the line

```python
print(re.findall(r"^The", "The dog barked"))
print(re.findall(r"^The", "Look, The dog"))
```

### `$` : end of the line
`\.` means a real period.

```python
print(re.findall(r"dog\.$", "I saw the dog."))
print(re.findall(r"dog\.$", "The dog. Yes"))

print("\n what the different and why??")
print(re.findall(r"dog.$", "I saw the dogs"))
```

### `\b` : word boundary
Finds *the* as a whole word, not inside *other*.

```python
text = "the other day"
print(re.findall(r"the", text))
print(re.findall(r"\bthe\b", text))
```

### `\b` with numbers
99 inside 299 is not a separate word.

```python
print(re.findall(r"\b99\b", "There are 99 bottles"))
print(re.findall(r"\b99\b", "There are 299 bottles"))
print(re.findall(r"\b99\b", "It costs $99"))
```

### `\B` : not a word boundary (inside a word)

```python
text = "the other mother"
print(re.findall(r"\Bthe\B", text))
```

## 5. Disjunction, grouping and precedence

### `|` : or

```python
text = "my cat and my dog"
print(re.findall(r"cat|dog", text))
```

### Why not `[catdog]`?
Brackets match only **one** character.

```python
text = "my cat"
print(re.findall(r"[catdog]", text))
```

### `( )` : grouping
`gupp(y|ies)` = gupp followed by y or ies.
With parentheses, `print` shows the part inside them: *y* comes from guppy, *ies* from guppies.

```python
text = "one guppy, two guppies, guppxxx"
print(re.findall(r"gupp(y|ies|xxx)", text))


print("\nAll word")
print([m.group(0) for m in re.finditer(r"gupp(y|ies|xxx)", text)])
```

### Without `( )`
`guppy|ies` means *guppy* or *ies*.

```python
text = "two guppies"
print(re.findall(r"guppy|ies", text))
```

### Precedence: counters before sequences
`the*` repeats only the **e**.

```python
text = "theeeee thethe th"
print(re.findall(r"the*", text))
```

### Precedence: sequences before `|`
`the|any` is *the* or *any*, never *thany*.

```python
text = "thany"
print(re.findall(r"the|any", text))
```

### Greedy `+` and non-greedy `+?`
Greedy takes as much as possible; non-greedy takes as little as possible.

```python
text = "once upon"
print(re.findall(r"[a-z]+", text))
print(re.findall(r"[a-z]+?", text))
print(re.findall(r"[a-z]", text))
```

```python
import re

text = "aaaa"

print(re.match(r"a+a", text).group())
print(re.match(r"a+?a", text).group())
```

## 6. A simple example: finding *the*

```python
text = "The other day the dog sat there."
print(re.findall(r"the", text))          # misses The, matches inside other and there
print(re.findall(r"[tT]he", text))       # finds The, still matches inside words
print(re.findall(r"\b[tT]he\b", text))   # correct
```

## 7. More operators

### `\d` : any digit (same as `[0-9]`)

### `\D` : any non-digit

```python
print(re.findall(r"\D", "a1b2"))
```

### `\w` : letter, digit or underscore

```python
print(re.findall(r"\w", "Hi_2!"))
```

### `\W` : not a letter, digit or underscore

```python
print(re.findall(r"\W", "Hi_2!?"))
```

### `\s` : whitespace

```python
print(re.findall(r"\s", "in Concord"))
```

### `\S` : non-whitespace

```python
print(re.findall(r"\S", "in Concord"))
```

### `\*` `\.` `\?` : special characters as normal characters
A backslash makes them literal.

```python
print(re.findall(r"\*", "K*A*P"))
print(re.findall(r"\.", "Dr. Livingston."))
print(re.findall(r"\?", "Why? How?"))
```

## 8. Substitutions and capture groups

### `re.sub` : replace
`re.sub(pattern, new, text)`

```python
text = "I like cherry pie"
print(re.sub(r"cherry", r"apricot", text))
```

### Capture groups `\1 \2 \3`
The `( )` save what matched. Here we swap month and day.

```python
text = "The date is 10/15/2011"
print(re.sub(r"(\d{2})/(\d{2})/(\d{4})", r"\2-\1-\3", text))
```

### `\1` to find a repeated word
`print` shows the word that was repeated.

```python
text = "I saw the the dog"
print(re.findall(r"\b([A-Za-z]+)\s+\1\b", text))

print("\n what the different !!")
text = "I saw the the dog"
print(re.findall(r"\b([A-Za-z]+)\s\1\b", text))
```

### `(?: )` : group without saving
The first pattern also saves `15/`; the second saves only the year. 
this **(?: )** is prevents the group captured only.

```python
print(re.findall(r"\d{2}/", "121/25/2026"))
```

```python
text = "The date is 10/15/2011"
print(re.findall(r"(\d{2}/){2}(\d{4})", text))
print(re.findall(r"(?:\d{2}/){2}(\d{4})", text))
```

## 9. ELIZA and lookahead

### ELIZA rule (from the book)

```python
text = "HE SAYS YOU ARE DEPRESSED MUCH OF THE TIME."
print(re.sub(r".* YOU ARE (DEPRESSED|SAD) .*", r"I AM SORRY TO HEAR YOU ARE \1", text))
```

### `(?= )` : true if the pattern comes next
Finds words followed by `?` without taking the `?`.

```python
text = "Is it ready? Who knows? Maybe."
print(re.findall(r"\w+(?=\?)", text))
```

### `(?! )` : true if the pattern does NOT come next
The first word, only if it does not start with t or T.

```python
print(re.findall(r"^(?![tT])\w+", "Hello world"))
print(re.findall(r"^(?![tT])\w+", "The world"))
```
