## Part 2 — Storing text in a list

A **string** holds one piece of text. A **list** holds many strings — for example, a small collection of sentences.

````python
sentences = [
    "Natural language processing is a field of computer science.",
    "Computers can understand human language.",
    "NLP is used in translation and chatbots.",
]

print(sentences)
````

### 2.1 Accessing items

````python
print("Number of sentences:", len(sentences))
print("First sentence     :", sentences[0])
print("Last sentence      :", sentences[-1])
````

### 2.2 Looping over the list

````python
for i, s in enumerate(sentences):
    print(i, s)
````

### 2.3 Adding a new sentence

````python
sentences.append("Python is a good language for NLP.")
print("Number of sentences now:", len(sentences))
````

## Part 3 — Getting text from the user

`input()` asks the user to type something and returns it as a string.
In VS Code a **text box appears at the top of the window** — type there and press **Enter**.

````python
user_text = input("Write a sentence: ")
print("You wrote:", user_text)
````

### 3.1 Collecting several sentences from the user into our list

````python
n = int(input("How many sentences do you want to add? "))

for i in range(n):
    s = input(f"Sentence {i + 1}: ")
    sentences.append(s)

print("\nAll sentences:")
for s in sentences:
    print("-", s)
````

---
## Part 4 — Simple text processing (two examples)

Text processing will have its own lecture. Here we only see two small examples.

### Example 1 — Counting words and characters
`split()` breaks a sentence into words at the spaces.

````python
for s in sentences:
    words = s.split()
    print(f"{len(words):>2} words | {len(s):>3} characters | {s}")
````

### Example 2 — Searching for a word
We convert both the sentence and the word to lowercase with `lower()` so that *NLP* and *nlp* match.

````python
keyword = input("Which word are you looking for? ")

found = 0
j=0
for s in sentences:
    if keyword.lower() in s.lower():
        print("Found in:"," ",j, s)
        found += 1
    j=j+1
print(f"\n'{keyword}' appears in {found} sentence(s).")
````

## Part 5 — Reading text from a .txt file

So far our text was written inside the code or typed by the user.
Real NLP work reads text from **files**.

### 5.1 Create the file
In the VS Code Explorer, right-click `labs/data` → **New File** → name it `sentences.txt`, and paste:
```
Natural language processing is a field of computer science.
Computers can understand human language.
NLP is used in translation and chatbots.
Students will build NLP tools step by step.
```
Save it (**Ctrl + S**).

> If you prefer, the next cell creates the same file with Python (`"w"` = write mode).

````python
text_to_save = '''Natural language processing is a field of computer science.
Computers can understand human language.
NLP is used in translation and chatbots.
Students will build NLP tools step by step.'''

with open("../data/sentences.txt", "w", encoding="utf-8") as f:
    f.write(text_to_save)

print("File saved.")
````

### 5.2 Read the whole file at once
- `open(path, "r")` opens the file for **reading**.
- `encoding="utf-8"` tells Python how the characters are stored in the file — always include it.
- `with` closes the file automatically when we finish.

````python
with open("../data/sentences.txt", "r", encoding="utf-8") as f:
     content = f.read()

print(content)
print("\nType:", type(content))
print("Characters:", len(content))
````

### 5.3 Read the file line by line into a list
Each line ends with a hidden newline character `\n`; `strip()` removes it.
Now the file becomes a **list of sentences** — exactly like Part 2.

````python
file_sentences = []

with open("../data/sentences.txt", "r", encoding="utf-8") as f:
    for line in f:
        file_sentences.append(line.strip())

print(file_sentences)
print("Number of lines:", len(file_sentences))
````

### 5.4 Apply our two examples to the file

````python
# Example 1 — words per line
for s in file_sentences:
    print(f"{len(s.split())} words | {s}")

# Example 2 — search nlp in each sentense
keyword = "nlp"
matches = [s for s in file_sentences if keyword in s.lower()]
print(f"\nLines containing '{keyword}': {len(matches)}")
for m in matches:
    print("-", m)
````
